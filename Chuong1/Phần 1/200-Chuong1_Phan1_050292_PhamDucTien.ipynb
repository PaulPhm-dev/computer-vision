```python
#Gọi thư viện
import os 
import shutil

# Đường dẫn tới thư mục ảnh:
path_original = 'images/tulips/Original/'

#Hiển thị danh sách file ảnh gốc.
print("Danh sách file ảnh:")
for count, filename in enumerate(os.listdir(path_original)): 
    print(count+1,')',filename)
```

```python
#Viết hàm copy_renamefile() sao chép và đổi tên các file ảnh từ một thư mục vào trong một thư mục khác

#Hàm xóa các file trong một thư mục:
def delete_files_in_directory(directory_path):
    if not os.path.exists(directory_path):
        print('Thư mục này không tồn tại!')
        return
    
    files = os.listdir(directory_path)
    for file in files:
        file_path = os.path.join(directory_path, file)
        if os.path.isfile(file_path):
            os.remove(file_path)
            
# Hàm sao chép và đổi tên file:
#Tham số đầu vào: 
# - path_old: đường dẫn thư mục chứa file ảnh gốc
# - path_new: đường dẫn thư mục chứa file ảnh mới
# - name_new: Tên file ảnh mới
def copy_renamefile(path_old, path_new, name_new): 
    try:
    #xóa tất cả các file cũ của thư mục lưu:
        delete_files_in_directory(path_new)
    #Thiết lập chỉ số cho tên file mới 
        for count, filename in enumerate(os.listdir(path_old)): 
        #Lấy tên file  cũ:
            src = path_old + filename 
        #thiết lập tên file mong muốn đổi:
            dst = path_new + name_new + str(count) + ".jpg"    
        # Sao chép và Thay thế file cũ bằng file mới 
            dest = shutil.copyfile(src, dst)
        print('Sao chép và đổi tên file thành công!')
    except OSError:
        print('Lỗi xử lý...!')
    
```

```python
#Thiết lập đường dẫn và tên file mới:
path_new = 'images/Tulips/New/'
name_new = 'Hoa_tulip_'
```

```python
#Gọi hàm copy và đổi tên file:
copy_renamefile(path_original, path_new, name_new)

#Hiển thị danh sách file sau khi đã đổi tên.
for count, filename in enumerate(os.listdir(path_new)): 
    print(count+1,')',filename)
print('Tổng số ảnh đã sao chép và đổi tên:', count+1)
```

```python
#HIỂN THỊ TOÀN BỘ ẢNH TRONG THƯ MỤC
import pathlib
import matplotlib.pyplot as plt
import cv2

#Lấy danh sách tên file có trong thư mục
pic_tulips = list(pathlib.Path(path_new).glob('*'))

#Thiết lập số hàng, số cột hiển thị ảnh
cot = 4
hang = len(pic_tulips)//cot + 1

#Hiển thị nhiều ảnh sử dụng Matplotlib:
plt.figure(figsize=(4*cot,4*hang))
for count, filename in enumerate(pic_tulips):
    #Đọc từng ảnh theo chế độ mặc định
    img = cv2.imread(str(filename))
    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    
    #Thiết lập khung hiển thị:
    plt.subplot(hang,cot,count+1)
    #Hiển thị ảnh
    plt.imshow(img)
    #lấy tên file hiển thị
    plt.title(str(filename).split("/")[-1])
    plt.grid(ls='--')
    plt.axis('on')
    
plt.show()
```

2. Sử dụng hàm sao chép và đổi tên các file ảnh đó theo cấu trúc NhomMH_MaSV_i.jpg, i là số tăng dần.
3. Hiển thị toàn bộ ảnh ra một khung đồ thị như ví dụ:


```python
#{SINH VIÊN VIẾT CODE VÀO ĐÂY}
import os
import shutil

# Cấu hình nhóm và mã sinh viên
NhomMH = "Nhom200"   # Thay thế bằng nhóm của bạn
MaSV = "2221050292"   # Thay thế bằng mã sinh viên của bạn

# Thư mục chứa ảnh gốc và thư mục lưu ảnh đã đổi tên
source_folder = "images/Root"
target_folder = "images/Renamed"

# Tạo thư mục đích nếu chưa có
os.makedirs(target_folder, exist_ok=True)

# Lấy danh sách file ảnh trong thư mục gốc
image_files = [f for f in os.listdir(source_folder) if f.endswith(".jpg")]

# Sao chép và đổi tên ảnh
for i, filename in enumerate(image_files, start=1):
    new_filename = f"{NhomMH}_{MaSV}_{i}.jpg"
    src_path = os.path.join(source_folder, filename)
    dst_path = os.path.join(target_folder, new_filename)
    shutil.copy(src_path, dst_path)
    print(f"Đã sao chép {filename} -> {new_filename}")




```

```python
import pathlib
import matplotlib.pyplot as plt
import cv2
path_new = 'images/Renamed'
#Lấy danh sách tên file có trong thư mục
pic_tulips = list(pathlib.Path(path_new).glob('*'))

#Thiết lập số hàng, số cột hiển thị ảnh
cot = 4
hang = len(pic_tulips)//cot + 1

#Hiển thị nhiều ảnh sử dụng Matplotlib:
plt.figure(figsize=(6*cot,5*hang))
for count, filename in enumerate(pic_tulips):
    #Đọc từng ảnh theo chế độ mặc định
    img = cv2.imread(str(filename))
    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    
    #Thiết lập khung hiển thị:
    plt.subplot(hang,cot,count+1)
    #Hiển thị ảnh
    plt.imshow(img)
    #lấy tên file hiển thị
    plt.title(str(filename).split("/")[-1])
    plt.grid(ls='--')
    plt.axis('on')
    
plt.show()
```
