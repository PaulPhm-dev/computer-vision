## 3. Một số hệ màu cơ bản trong xử lý ảnh
---
Không gian màu (Colour Space) – Hệ màu: được hiểu là các mô hình toán để miêu tả màu sắc. Hệ thống không gian màu cho phép mỗi mà được xác định theo số học. Mỗi không gian màu đều có một tác dụng và ứng dụng trong các bài toán khác nhau.Một số hệ màu thông dụng:

* Hệ màu RGB (Red – Green – Blue)
* Hệ màu RGBA
* Hệ màu YUV
* Hệ màu CMYK
* Hệ màu HSV (Hue, Saturation, Value)
* Hệ màu HSL (Hue, Staturation, Luminance)
* …


### 3.1 Hệ màu RGB
---
* RGB là không gian màu phổ biến dùng trong máy tính, máy ảnh, điện thoại và nhiều thiết bị kĩ thuật số khác nhau. 
* Không gian màu này khá gần với cách mắt người tổng hợp màu sắc. 
* Nguyên lý cơ bản là sử dụng 3 màu sắc cơ bản R (red - đỏ), G (green - xanh lục) và B (blue - xanh lam) để biểu diễn tất cả các màu sắc.
* Thông thường, trong mô hình 24 bit mỗi kênh màu sẽ sử dụng 8bit để biểu diễn, tức là giá trị R, G, B nằm trong khoảng 0 - 255. Bộ 3 số này biểu diễn cho từng điểm ảnh, mỗi số biểu diễn cho cường độ của một màu. Với mô hình màu 24bit thì số màu tối đa có thể tạo ra là 255 x 255 x 255 = 16581375 màu.

2. Hệ màu CMYK

```python
import cv2
import numpy as np
from PIL import Image

def rgb_to_cmyk(img_RGB):
    # Chuyển đổi ảnh sang kiểu float32 và chuẩn hóa về [0,1]
    img_rgb = img_RGB.astype(np.float32) / 255.0
    
    # Tách kênh R, G, B
    R, G, B = cv2.split(img_RGB)

    # Tính kênh K (Black)
    K = 1 - np.max([R, G, B], axis=0)

    # Tránh lỗi chia cho 0 khi K = 1
    C = (1 - R - K) / (1 - K + 1e-8)
    M = (1 - G - K) / (1 - K + 1e-8)
    Y = (1 - B - K) / (1 - K + 1e-8)

    # Chuyển đổi về dạng uint8 (0-255)
    C, M, Y, K = [np.uint8(channel * 255) for channel in [C, M, Y, K]]

    return C, M, Y, K


# Chuyển đổi sang CMYK
C, M, Y, K = rgb_to_cmyk(img_RGB)

# Tạo ảnh CMYK bằng Pillow
img_cmyk = Image.merge("CMYK", [Image.fromarray(C), Image.fromarray(M), Image.fromarray(Y), Image.fromarray(K)])

# Lưu ảnh CMYK
img_cmyk.save("image_cmyk.jpg")

# Hiển thị giá trị kênh CMYK của pixel đầu tiên
print(f"C: {C[0, 0]}, M: {M[0, 0]}, Y: {Y[0, 0]}, K: {K[0, 0]}")
plt.figure(figsize=(20,8))
plt.subplot(1,2,2)
plt.title("Hệ màu CMYK ")
plt.imshow(img_cmyk)


#2. Hệ màu HSV
img_HSV = cv2.cvtColor(img_RGB,cv2.COLOR_BGR2HSV)
plt.subplot(1,2,1)
plt.title('hệ màu HSV')
plt.imshow(img_HSV)
#3.Sang Hệ YUV
plt.show()
```
