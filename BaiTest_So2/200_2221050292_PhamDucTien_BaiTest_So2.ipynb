```python
#Khai báo sử dụng thư viện:
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

# Bài 1

# Yêu cầu 1.1:

```python
#đọc ảnh chế độ ảnh xám 
img_pic1 = cv2.imread('Pic/pic1.jpg',0)

#hiển thị ảnh 
plt.imshow(img_pic1 , cmap = 'gray')
plt.title('Ảnh gốc img_pic1')
plt.show()
#độ phân giải của ảnh 
h = img_pic1.shape[0]
w = img_pic1.shape[1]
print('1.Độ phân giải của ảnh : ' , h , 'x' , w)
print('2.Số pixel điểm ảnh  : ' , h * w)
```

```python
plt.figure(figsize = (20,15))
plt.subplot(1,2,1) 
plt.imshow(img_pic1 , cmap = 'gray') 
plt.title('Ảnh gốc') 

plt.subplot(1,2,2) 
plt.hist(img_pic1.ravel(),bins = 256) 
plt.title('Histogram của ảnh') 
plt.show()
```

```python
anh = img_pic1.mean()

if anh > (255 / 2):
    print('→ Ảnh SÁNG')
else:
    print('→ Ảnh TỐI')
```

# Yêu cầu 1.2:

```python
img_cb = cv2.equalizeHist(img_pic1)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_pic1, cmap='gray')
plt.title('Ảnh gốc')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_cb, cmap='gray')
plt.title('Ảnh đã cân bằng sáng')
plt.axis('off')

plt.show()
```

```python
cv2.imwrite('Pic/Saves/2221050292_Equallization1.jpg', img_cb)
```

# Yêu cầu1.3:

```python
clahe = cv2.createCLAHE(clipLimit=6, tileGridSize=(20, 30))
img_cb2 = clahe.apply(img_pic1)
plt.figure(figsize = (12,5))
plt.subplot(1,2,1) 
plt.imshow(img_cb2 , cmap = 'gray') 
plt.title('Ảnh đã cân bằng') 

plt.subplot(1,2,2) 
plt.hist(img_cb2.ravel(),bins = 256) 
plt.title('Histogram của ảnh đã cân bằng') 
plt.show()
```

```python
cv2.imwrite('Pic/Saves/2221050292_Equallization2.jpg', img_cb)
```

# Bài 2

# Yêu cầu 2.1

```python
#đọc ảnh chế độ ảnh xám 
img_pic2 = cv2.imread('Pic/pic2.jpg',0)

#hiển thị ảnh 
plt.imshow(img_pic2 , cmap = 'gray')
plt.title('Ảnh gốc img_pic2')
plt.show()
#độ phân giải của ảnh 
h2 = img_pic2.shape[0]
w2 = img_pic2.shape[1]
print('1.độ phân giải của ảnh : ' , h2 , 'x' , w2)
print('2.số pixel điểm ảnh  : ' , h2 * w2)
```

```python
#thực hiện đảo ảnh 
img_dao = img_pic2.max() - img_pic2 


#Hiển thị kết quả ảnh trước và sau khi đảo ảnh:
plt.figure(figsize = (12,5))

plt.subplot(1,2,1) 
plt.imshow(img_pic2,cmap = 'gray') 
plt.title('anh goc') 
#Hiển thị kết quả ảnh sau và sau khi đảo ảnh:
plt.subplot(1,2,2) 
plt.imshow(img_dao,cmap = 'gray') 
plt.title('ảnh đã xử lý đảo') 

plt.show()
#lưu ảnh 
cv2.imwrite('Pic/Saves/2221050292_hand_inv.jpg', img_dao)
```

# Yêu cầu 2.3

```python
#Thực hiện biến đổi ảnh với hàm mũ:
c = 12
a = 1.5
img_pic2_gama = c*pow(img_pic2,a)
img_pic2_gama = img_pic2_gama.astype(np.int32) 


#Hiển thị kết quả trước và sau khi biến đổi:
plt.figure(figsize = (12,10)) 
plt.subplot(1,2,1) 
plt.imshow(img_pic2,cmap= 'gray') 
plt.title('Ảnh gốc ban đầu') 
plt.subplot(1,2,2) 


plt.imshow(img_pic2_gama,cmap= 'gray') 
plt.title('Ảnh đã xử lý bằng hàm (c = 12 , a = 1.5 )') 
plt.show()

#lưu ảnh  
cv2.imwrite('Pic/Saves/2221050292_hand_gamma.jpg', img_pic2_gama)
```

# Yêu cầu 2.4

```python
kernel =  np.array(    [[-5,-8,-10,-8,-5],
                        [-4,-10,-20,-10,-4],
                        [0,0,0,0,0],
                        [4,10,20,10,4],
                        [5,8,10,8,5]]
                       )

print( kernel)
```

```python
#Thực hiện lọc ảnh với kernel ở trên:
img_kernel = cv2.filter2D(img_dao, -1 , kernel) 
#Hiển thị kết quả 
plt.figure(figsize = (10,5))

plt.subplot(1,2,1)
plt.imshow(img_dao,cmap = 'gray')
plt.title('ảnh gốc') 

plt.subplot(1,2,2)
plt.imshow(img_kernel,cmap = 'gray')
plt.title('ảnh qua bộ lọc kernel ') 
plt.show()
#lưu ảnh 
cv2.imwrite('Pic/Saves/2221050292_hand_2D.jpg', img_kernel)
```

# Bài 3:
# Yêu cầu 3.1:

```python

img = cv2.imread('Pic/Bai3/Test_06.jpg')
#độ phân giải của ảnh 
h = img.shape[0]
w = img.shape[1]
print('1.Độ phân giải của ảnh : ' , h , 'x' , w)
print('2.Số pixel điểm ảnh  : ' , h * w)
img_RGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
#LẤY DỮ LIỆU CÁC KÊNH MÀU:
B = img_RGB.copy()
B[:, :, [1,2]] = 0 

G = img_RGB.copy()
G[:, :, [0, 2]] = 0 

R = img_RGB.copy()
R[:, :, [0, 1]] = 0 

# Hiển thị từng kênh
plt.figure(figsize=(10, 4))

plt.subplot(1, 4, 1)
plt.imshow(img_RGB)
plt.title('Ảnh RGB')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(R, cmap='Reds')
plt.title('Red channel')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(G, cmap='Greens')
plt.title('Green channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(B, cmap='Blues')
plt.title('Blue channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# Yêu cầu 3.2

```python
# 1 chuyển ảnh xám 
img_xam = cv2.cvtColor(img_RGB, cv2.COLOR_RGB2GRAY)

nguong, img_dentrang = cv2.threshold(img_xam, 128, 255, cv2.THRESH_BINARY)


plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(img_RGB)
plt.title('Ảnh màu')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_xam , cmap = 'gray')
plt.title('Ảnh xám')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_dentrang ,cmap = 'gray')
plt.title('Ảnh đen trắng')
plt.axis('off')
#Lưu ảnh  
cv2.imwrite('Pic/Saves/2221050292_gray.png', img_xam)
cv2.imwrite('Pic/Saves/2221050292_binary.png', img_dentrang)
```

# Yêu cầu 3.3
