```python
#Import các thư viện 
import numpy as np 
import matplotlib.pyplot as plt 
import cv2 
```

# Yêu	cầu	1.1:

```python
# Đọc ảnh ở chế độ màu (BGR mặc định của OpenCV)
image = cv2.imread('Pic/Pic_T1.png', cv2.IMREAD_COLOR)

# Chuyển sang RGB để hiển thị đúng màu
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Hiển thị ảnh
plt.imshow(image)
plt.title('Ảnh màu RGB')
plt.show()
```

# Yêu cầu 1.2:

```python
# Độ phân giải của ảnh (h x w) và số pixel điểm ảnh
h, w = image.shape[:2]  
total_pixels = h * w  

print(f"Độ phân giải của ảnh: {h} x {w} (h x w)")
print(f"Tổng số pixel điểm ảnh: {total_pixels}")
```

# Yêu cầu 1.3

```python

#Resize ảnh với các thuật toán khác nhau:

img_400 = cv2.resize(image , (400,400), cv2.INTER_AREA)

plt.imshow(img_400)
plt.title('Ảnh thu nhỏ 400 x 400')
plt.show()
# cho biết độ phân giải của ảnh (h x w) và số pixel điểm ảnh
h2, w2 = img_400.shape[:2]  
total_pixels2 = h2 * w2  
print(f"Ảnh sau khi thay đổi kích thước :")
print(f"Độ phân giải của ảnh: {h2} x {w2} (h2 x w2)")
print(f"Tổng số pixel điểm ảnh: {total_pixels2}")
```

# Yêu cầu 1.4

```python
img_gray = cv2.cvtColor(img_400, cv2.COLOR_BGR2GRAY)
#Chuyển sang ảnh nhị phân (0 hoặc 255)
_, binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)
#Sử dụng phương thức Laplacian của OpenCV:
img_lap_cv = cv2.Laplacian(binary, -1, 5)


#Hiển thị kết quả:
plt.figure(figsize=(15,8))
plt.subplot(1,2,1)
# Hiển thị ảnh đen trắng
plt.imshow(binary, cmap='gray')
plt.title('Ảnh đen trắng')


plt.subplot(1,2,2)
plt.imshow(img_lap_cv, cmap = 'gray')
plt.title('Lấy biên theo laplacian của openCV')
plt.show()
#lưu ảnh 
cv2.imwrite('Pic/2221050292_Edge.jpg', img_lap_cv)
```

```python
#Sử dụng phương thức canny:
#Thực hiện làm mịn ảnh với Gausian:
blur = cv2.GaussianBlur(binary, (5,5), 0)
img_canny = cv2.Canny(blur, 8,170,5)

#Hiển thị kết quả:
plt.figure(figsize=(15,5))
plt.subplot(1, 3, 2)
plt.imshow(img_canny,cmap='gray')
plt.title('Biên với Canny')

plt.show()
```

# Bài 3:

```python

#Bước 1: Đọc ảnh ở chế độ ảnh màu:
image_2 = cv2.imread('Pic/Pic_T3.jpg')

#Hiển thị ảnh gốc ở dạng xám:
gray_2 = cv2.cvtColor(image_2, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.imshow(gray_2, cmap='gray');
plt.title('Ảnh gốc')

#Bước 2: Phát hiện biên với Canny:
#Thực hiện làm mịn ảnh với Gausian:
blur_2 = cv2.GaussianBlur(gray_2, (11, 11), 0)

#Phát hiện biên với Canny
canny_2 = cv2.Canny(blur_2, 90, 110, 2)
plt.subplot(2,2,2)
plt.imshow(canny_2, cmap='gray')
plt.title('Ảnh biên với Canny')


#Bước 3: Làm đậm đường biên:
dilated = cv2.dilate(canny_2, (1,1), iterations = 2)
plt.subplot(2,2,3)
plt.imshow(dilated, cmap='gray')
plt.title('Ảnh làm đậm đường biên')

#Bước 4: Lấy countour của ảnh và đếm số lượng:
(cnt, heirarchy) = cv2.findContours(dilated.copy(), 
                                    cv2.RETR_EXTERNAL, 
                                    cv2.CHAIN_APPROX_NONE)

rgb_2 = cv2.cvtColor(image_2, cv2.COLOR_BGR2RGB)
cv2.drawContours(rgb_2, cnt, -1, (0,255,0), 2)

plt.subplot(2,2,4)
plt.imshow(rgb_2)
plt.title('Ảnh chồng lấp countour')
plt.show()

print('Số lượng quả bóng trong ảnh là: ', len(cnt))
```

# Bài 4

```python
img_color = cv2.imread('Pic/Pic_T4.jpg')
img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)
img = img_color.copy()

#lấy kích thước ảnh mẫu:
h = img_color.shape[0]
w = img_color.shape[1]
print(img_color.shape)
print('Chiều cao:', h)
print('Chiều rộng:', w)



plt.figure(figsize=(10, 6))
plt.imshow(img_color)
plt.title('Ảnh gốc')
plt.grid(True, color='red', linestyle='--', alpha=0.5)
plt.show()
```

```python
regions = [(600, 690, 550, 650),
           (650, 720, 870, 970),
           (1030, 1180, 1330, 1490),]

for i, (y1, y2, x1, x2) in enumerate(regions):
    template = img_gray[y1:y2, x1:x2]
    h, w = template.shape

    result = cv2.matchTemplate(img_gray, template, cv2.TM_CCOEFF_NORMED)
    _, max_val, _, max_loc = cv2.minMaxLoc(result)

    top_left = max_loc
    bottom_right = (top_left[0] + w, top_left[1] + h)
    
    cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 5)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.title('Hiển thị kết quả tìm kiếm trong ảnh gốc')
plt.imshow(img)

plt.subplot(1, 2, 2)

plt.title('Heatmap - Vùng trùng khớp')
plt.imshow(result, cmap='jet')
plt.show()
```
