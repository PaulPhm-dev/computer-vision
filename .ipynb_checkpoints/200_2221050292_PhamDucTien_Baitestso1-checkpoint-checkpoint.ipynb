### 1.1 Thay đổi kích thước ảnh (Resizing images)
---
* Thay đổi kích thước (Zoom in, Zoom out) là một trong những kỹ thuật xử lý ảnh thông dụng.
* Ảnh có kích thước lớn sẽ chiếm nhiều bộ nhớ, số lượng tập dữ liệu ảnh thường rất lớn. Do đó cần giảm kích thước ảnh. Resize là một kỹ thuật phổ biến. => Trong OpenCV, Sử dụng hàm cv2.resize()

```python
#Đọc và hiển thị ảnh gốc ban đầu: Apples.jpg
img_original = cv2.imread('images/Apples.jpg')

img_original_RGB = cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB)

plt.imshow(img_original_RGB)

plt.title('Ảnh gốc')

plt.show()

```

```python
#Lấy kích thước ảnh ban đầu:
h = img_original_RGB.shape[0]
w = img_original_RGB.shape[1]
print(h , 'x' , w)
```

```python
# Thay đổi kích thước ảnh
#Thu nhỏ ảnh
h1 = int(h//2)
w1 = int(w//2)
img_zoom_in = cv2.resize(img_original_RGB, (w1,h1))
#Phóng to ảnh
h2 = h*2
w2 = w*2
img_zoom_out = cv2.resize(img_original_RGB, (w2,h2))

plt.figure(figsize= (10,5))
plt.subplot(1,2,1)
plt.imshow(img_zoom_in)
plt.title('Ảnh đã thu nhỏ (giảm 50% - 2221050436)')


plt.subplot(1,2,2)
plt.imshow(img_zoom_out)
plt.title('Ảnh đã phóng to (giảm 200% - 2221050436)')

plt.show()
print('Độ phân giải của ảnh: ', img_zoom_in.shape)
print('Số phần tử của ảnh: ', img_zoom_in.size)

print('--------------------------------------')

print('Độ phân giải của ảnh: ', img_zoom_out.shape)
print('Số phần tử của ảnh: ', img_zoom_out.size)

# Lưu ảnh
cv2.imwrite('images/Apples_zoom_in.jpg', cv2.cvtColor(img_zoom_in, cv2.COLOR_RGB2BGR)) #chuyển đổi không gian của màu ảnh 
cv2.imwrite('images/Apples_zoom_out.jpg', cv2.cvtColor(img_zoom_out, cv2.COLOR_RGB2BGR)) 
```

## Các tham số của phương thức resize:
---

**cv2.resize(src, dsize, interpolation)**

Ý nghĩa các tham số:

1. src: ảnh gốc cần thay đổi kích thước.
2. dsize (width, height): Kích thước ảnh muốn thay đổi
3. interpolation: chỉ định thuật toán thực hiện resize:
* INTER_NEAREST – nội suy láng giềng gần nhất
* INTER_LINEAR – nội suy song tuyến tính (mặc định)
* INTER_AREA – nội suy theo vùng.
* INTER_CUBIC –  nội suy xoắn bậc 3, trên  4×4 pixel láng giềng
* INTER_LANCZOS4 –  nội suy Lanczos trên 8×8 pixel láng giềng!


**Thông thường:**
* cv.INTER_AREA để thu nhỏ
* cv.INTER_CUBIC & cv.INTER_LINEAR để zoom.

```python
#Resize ảnh với các thuật toán khác nhau:

img_apple_256 = cv2.resize(img_original_RGB , (256,256), cv2.INTER_AREA)

plt.imshow(img_apple_256)
plt.title('Ảnh thu nhỏ 256 x 256')
plt.show()
```

```python
img_apple_2000 = cv2.resize(img_original_RGB , (2000,2000), cv2.INTER_CUBIC)

plt.imshow(img_apple_2000)
plt.title('Ảnh thu nhỏ 2000 x 2000')
plt.show()
```

```python
#Thay đổi kích thước ảnh theo tỷ lệ % (scale) nhập vào: 
# Nếu scale <100 thu nhỏ 
# Nếu scale = 100 giữ nguyên
# Nếu scale>100 phóng to

tyle = int(input('Nhập vào tỷ lệ muốn thay đổi(%): '))

h_tyle = int(h * tyle/100)
w_tyle = int(w * tyle/100)

st1 = 'Kích thước ảnh gốc: ' +  str(h) + "x" +  str(w)
st2 = 'Kích thước ảnh thay đổi: ' +  str(h_tyle) + "x" +  str(w_tyle)
print(st1)
print(st2)
print('Kích thước ảnh sau thay đổi theo tỷ lệ: ', tyle, '% là :', h_tyle, 'x', w_tyle)

img_scale = cv2.resize(img_original_RGB, (w_tyle, h_tyle), cv2.INTER_NEAREST)

# Hiển thị ảnh
plt.figure(figsize= (15,5))
plt.subplot(1,2,1)
plt.imshow(img_original_RGB)
plt.title('st1')


plt.subplot(1,2,2)
plt.imshow(img_scale)
plt.title('st2')

plt.show()
```

## THỰC HÀNH 2.1 (Tiếp): 
---
Thay đổi kích thước nhiều ảnh:

**Yêu cầu 4:** Sinh viên sử dụng thư mục chứa các ảnh đã đổi tên thực hiện trong bài thực hành 1.3 chương 1. Thực hiện thay đổi kích thước toàn bộ ảnh về độ phân giải 256x256 pixel. Lưu lại ảnh đã thay đổi vào thư mục mới có tên Pic_256 và đặt lại tên ảnh như sau: Pic_256_i.jpg với i là các số thứ tự của ảnh.

```python
import cv2
import os

# Thư mục nguồn chứa ảnh gốc
input_folder = "images/Renamed/"

# Thư mục đích để lưu ảnh đã resize
output_folder = "Pic_256"

# Tạo thư mục nếu chưa tồn tại
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Lấy danh sách file ảnh trong thư mục nguồn
image_files = sorted(os.listdir(input_folder))  # Sắp xếp để đánh số thứ tự đúng

# Duyệt qua từng ảnh và resize
for i, filename in enumerate(image_files, start=1):
    input_path = os.path.join(input_folder, filename)  # Đường dẫn file gốc
    output_path = os.path.join(output_folder, f"Pic_256_{i}.jpg")  # Đường dẫn file mới

    # Đọc ảnh
    img = cv2.imread(input_path)

    if img is not None:  # Kiểm tra nếu ảnh hợp lệ
        # Resize về 256x256
        resized_img = cv2.resize(img, (256, 256))

        # Lưu ảnh đã resize vào thư mục Pic_256
        cv2.imwrite(output_path, resized_img)

print(" Resize hoàn tất! Ảnh đã lưu trong thư mục 'Pic_256'.")
```

```python
import os
import cv2
import matplotlib.pyplot as plt

# Thư mục chứa ảnh
image_folder = "Pic_256"

# Lấy danh sách tất cả file ảnh (lọc theo định dạng JPG, PNG)
image_files = sorted([f for f in os.listdir(image_folder) if f.endswith(('.jpg', '.png'))])

# Số lượng ảnh
num_images = len(image_files)

# Xác định số hàng và cột để hiển thị (ví dụ: mỗi hàng có 5 ảnh)
cols = 5
rows = (num_images // cols) + (num_images % cols > 0)  # Tính số hàng linh hoạt

# Tạo figure
plt.figure(figsize=(15, rows * 3))  # Điều chỉnh kích thước theo số hàng

# Duyệt qua danh sách ảnh và hiển thị
for i, filename in enumerate(image_files):
    img_path = os.path.join(image_folder, filename)  # Đường dẫn ảnh
    img = cv2.imread(img_path)  # Đọc ảnh với OpenCV

    if img is not None:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Chuyển BGR → RGB để hiển thị đúng màu
        plt.subplot(rows, cols, i + 1)  # Vị trí subplot
        plt.imshow(img)
        plt.title(f"Ảnh {i+1}")  # Tiêu đề ảnh
        plt.axis("off")  # Ẩn trục tọa độ

plt.tight_layout()  # Căn chỉnh khoảng cách giữa các ảnh
plt.show()  # Hiển thị ảnh
```

## 2. Cắt ảnh (Cropping Images)
---
Thực hiện cắt một phần của ảnh, sử dụng phương pháp tách các phần tử trong ma trận:
* Ảnh nhị phân, ảnh xám --> Ma trận [hàng (chiều cao ảnh), côt (chiều rộng ảnh)]
* Ảnh màu ---> Mảng 3 chiều [hàng (chiều cao ảnh), côt (chiều rộng ảnh),màu (BGR)]

**img_cropped = img[start_row:end_row, start_col:end_col]**

```python
#Đọc và hiển thị ảnh màu: MissVN.jpg
img_crop = cv2.imread('images/MissVN.jpg')

img_crop_RGB = cv2.cvtColor(img_crop,cv2.COLOR_BGR2RGB)

# Tạo 1 vecto khoảng cách 20
ax = np.arange(0, img_crop.shape[1], 20)
ay = np.arange(0, img_crop.shape[0], 20)

plt.figure(figsize=(10,6))
plt.imshow(img_crop_RGB)
plt.title('ẢNH HHOA HẬU VIỆT NAM')
plt.xticks(ax, rotation=90) # Thiết lập chia trục X
plt.yticks(ay) # Thiết lập chia trục Y
plt.grid(ls='--')
plt.show()


```

```python
#Cắt lấy từng người trong ảnh gốc:
img_ahau2 = img_crop_RGB[40:,20:220]

img_hoahau = img_crop_RGB[10:, 220:420]

img_ahau1 = img_crop_RGB[40:,425:625]

plt.figure(figsize=(15,6))
plt.subplot(1,3,1)
plt.imshow(img_hoahau)
plt.title('Hoa hậu')

plt.subplot(1,3,2)
plt.imshow(img_ahau1)
plt.title('Á hậu 1')

plt.subplot(1,3,3)
plt.imshow(img_ahau2)
plt.title('Á hậu 2')

plt.show()
```

```python
#CẮT ẢNH THEO MỘT VÙNG LỰA CHỌN BẰNG CHUỘT VỚI PHƯƠNG THỨC cv2.selectROI():
import cv2
import numpy as np
import matplotlib.pyplot as plt  
  
#Đọc ảnh:
img_miss = cv2.imread('images/MissVN.jpg')
  
# Select ROI: Lấy một vùng trong ảnh --> Toạ độ của vùng ảnh 
r = cv2.selectROI("select the area", img_miss)

print(r)

#Cắt ảnh theo vùng được chọn
cropped_image = img_miss[int(r[1]):int(r[1]+r[3]), 
                      int(r[0]):int(r[0]+r[2])]
  
# Hiển thị ảnh đã cắt:
plt.imshow(cv2.cvtColor(cropped_image,cv2.COLOR_BGR2RGB))
plt.grid(ls='--')
plt.show()
```

Hàm cv2.selectROI() trả về một tuple gồm 4 giá trị: (x,y,w,h)

* x: Tọa độ x của góc trên bên trái của hình chữ nhật (tính từ góc trái trên cùng của ảnh).
* y: Tọa độ y của góc trên bên trái của hình chữ nhật (tính từ góc trái trên cùng của ảnh).
* w: Chiều rộng (width) của hình chữ nhật.
* h: Chiều cao (height) của hình chữ nhật.

```python
#làm mờ một phần của ảnh được chọn:
#Đọc ảnh:
img_miss = cv2.imread('images/MissVN.jpg')
  
# Select ROI: Lấy một vùng trong ảnh --> Toạ độ của vùng ảnh 
r = cv2.selectROI("select the area", img_miss)

print(r)

#Đổi màu theo vùng được chọn
img_miss[int(r[1]):int(r[1]+r[3]), 
                      int(r[0]):int(r[0]+r[2])] = 100
  
# Hiển thị ảnh đã cắt:
plt.imshow(cv2.cvtColor(cropped_image,cv2.COLOR_BGR2RGB))
plt.grid(ls='--')
plt.show()
```

## 3. Xoay ảnh (Rotate Images)
---
OpenCV hỗ trợ phương thức xoay ảnh:
 **cv2.rotate(img, rotateCode)**
 
 Trong đó:
 1. img: ảnh gốc muốn xoay
 2. rotateCode: các chế độ xoay ảnh:
     * cv2.ROTATE_90_CLOCKWISE: Xoay 90 độ theo chiều kim đồng hồ
     * cv2.ROTATE_90_COUNTERCLOCKWISE: Xoay 90 độ ngược chiều kim đồng hồ
     * cv2.ROTATE_180: Xoay ảnh 180 (lật ảnh)

```python
#Đọc và hiển thị ảnh màu: Pic_1.jpg
img_rotate = cv2.imread('images/Pic_1.jpg')

#Chuyển hệ màu:
img_rotate_rgb = cv2.cvtColor(img_rotate, cv2.COLOR_BGR2RGB)

#Hiển thị
plt.figure(figsize=(4,7))
plt.imshow(img_rotate_rgb)
plt.title('Ảnh gốc')
plt.show()





```

```python
#Xoay ảnh 90 độ theo chiều kim đồng hồ: (cv2.ROTATE_90_CLOCKWISE)
img_rotate1 = cv2.rotate(img_rotate_rgb, cv2.ROTATE_90_CLOCKWISE)

#Hiển thị ảnh:

plt.imshow(img_rotate1)
plt.title('Ảnh xoay 90 độ')
plt.show()

```

```python
#Xoay ảnh 90 độ ngược chiều kim đồng hồ: (cv2.ROTATE_90_COUNTERCLOCKWISE)
img_rotate2 = cv2.rotate(img_rotate_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)

#Hiển thị ảnh:

plt.imshow(img_rotate2)
plt.title('Ảnh xoay 90 độ')
plt.show()



```

```python
#Xoay ảnh 180 độ: (cv2.ROTATE_180)
img_rotate3 = cv2.rotate(img_rotate_rgb, cv2.ROTATE_180)

#Hiển thị ảnh:

plt.imshow(img_rotate3)
plt.title('Ảnh xoay 180 độ')
plt.show()

```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,4,1)
plt.imshow(img_rotate_rgb,)
plt.title('Ảnh gốc')

plt.subplot(1,4,2)
plt.imshow(img_rotate1)
plt.title('xoay 90 độ cùng chiều kim đồng hồ')

plt.subplot(1,4,3)
plt.imshow(img_rotate2)
plt.title('xoay 90 độ ngược chiều kim đồng hồ')

plt.subplot(1,4,4)
plt.imshow(img_rotate3)
plt.title('xoay 180 độ')
plt.show()
```

## 4. Lật ảnh (Flip Images)
---
OpenCV hỗ trợ phương thức lật ảnh:
 **cv2.flip(img, flipCode)**
 
 Trong đó:
 1. img: ảnh gốc muốn xoay
 2. flipCode: các chế độ lật ảnh:
     * flipCode = 0: Lật ảnh theo chiều ngang
     * flipCode > 0: Lật ảnh theo chiều dọc
     * flipCode < 0: Lật ảnh theo chiều ngang và dọc

```python
#Đọc và hiển thị ảnh màu: fish.jpg
import cv2
import matplotlib.pyplot as plt
img_flip = cv2.imread('images/fish.jpg')

img_flip_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,5))
plt.imshow(img_flip_rgb)
plt.title("Ảnh gốc")
plt.show()

```

```python
#Lật ảnh theo chiều ngang (trục X): flipCode = 0
img_flip_x = cv2.flip(img_flip_rgb,0)
#Hiển thị
plt.figure(figsize=(4,7))
plt.imshow(img_flip_x)
plt.title('Ảnh lật theo trục x')
plt.show()
```

```python
#Lật ảnh theo chiều dọc (trục Y): flipCode > 0
img_flip_y = cv2.flip(img_flip_rgb,1)

plt.figure(figsize=(4,7))
plt.imshow(img_flip_y)
plt.title('ẢNh lật theo trục x')
plt.show()
```

```python
#Lật ảnh theo cả 2 chiều ngang, dọc (theo trục Y, X): flipCode < 0
img_flip_yx = cv2.flip(img_flip_rgb,-1)

plt.figure(figsize=(4,7))
plt.imshow(img_flip_yx)
plt.title('Ảnh lật theo trục yx')
plt.show()
```

```python
plt.figure(figsize=(12,10))
plt.subplot(2,2,1)
plt.imshow(img_flip_rgb)
plt.title('Ảnh gốc')

plt.subplot(2,2,2)
plt.imshow(img_flip_x)
plt.title('lật theo trục x')

plt.subplot(2,2,3)
plt.imshow(img_flip_y)
plt.title('lật theo trục y')

plt.subplot(2,2,4)
plt.imshow(img_flip_yx)
plt.title('lật theo trục yx')
plt.show()
```

## THỰC HÀNH 2.3 (Tiếp): 
---
Thay đổi kích thước nhiều ảnh:

**Yêu cầu 2:** Sinh viên sử dụng thư mục Pic_256 đã tạo ở bài thực hành 2.1 yêu cầu 4 ở trên. với mỗi một ảnh trong thư mục thực hiện lật ảnh theo trục Y và lưu thành một ảnh mới vào cùng thư mục Pic_256 với tên ảnh là tên ảnh cũ + flip ví dụ: Pic_256_1_flip.jpg

```python
#{SINH VIÊN VIẾT CODE VÀO ĐÂY}
import pathlib
import matplotlib.pyplot as plt
import cv2
import os

# Định nghĩa đường dẫn
path_original = 'images/New/'
path_save = 'images/Pic_256/'

# Tạo thư mục lưu ảnh nếu chưa tồn tại
os.makedirs(path_save, exist_ok=True)

# Lấy danh sách tên file có trong thư mục
pic_tulips = list(pathlib.Path(path_original).glob('*'))

# Thiết lập số hàng, số cột hiển thị ảnh
cot = 4
hang = len(pic_tulips) // cot + 1

plt.figure(figsize=(3 * cot, 3 * hang))

for i, filename in enumerate(pic_tulips, start=1):  # Bắt đầu từ 1
    # Đọc ảnh
    img = cv2.imread(str(filename))
    if img is None:
        continue  # Bỏ qua nếu không đọc được ảnh

    # Resize ảnh về 256x256
    img = cv2.resize(img, (256, 256))

    # Lật ảnh ngang
    img = cv2.flip(img, 1)

    # Lưu ảnh với tên mới theo định dạng "Pic_256_{i}_flip.jpg"
    new_filename = f"Pic_256_{i}_flip.jpg"
    save_path = os.path.join(path_save, new_filename)
    cv2.imwrite(save_path, img)

    # Chuyển ảnh từ BGR sang RGB để hiển thị đúng màu
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Thiết lập khung hiển thị
    plt.subplot(hang, cot, i)
    plt.imshow(img_rgb)
    plt.axis('off')  # Ẩn trục tọa độ

# Hiển thị ảnh
plt.tight_layout()
plt.show()

print(f"Đã lưu {len(pic_tulips)} ảnh vào thư mục '{path_save}' với tên định dạng 'Pic_256_{i}_flip.jpg'")



```
