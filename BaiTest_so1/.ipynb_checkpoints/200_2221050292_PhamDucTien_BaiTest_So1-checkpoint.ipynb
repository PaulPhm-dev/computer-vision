**Yêu câu 3:** Thực hiện thay đổi kích thước của ảnh gốc theo tham	số tỷ lệ % nhập vào từ bàn phím.Nếu	số	nhập vào < 100 (thu	nhỏ) sử	dụng thuật toán cv2.INTER_LANCZ0S4,	ngược	lại	nếu	số	nhập	vào	>=100	(phóng	to)	sử	dụng thuật toán cv2.INTER_AREA.	Hiển thị kết quả như hình minh họa dưới đây và lưu ảnh đã	thay	đổi	vào	thư	mục	Pic/Saves/	trong	đó	MaSV_Scale_In.jpg	nếu	ảnh	thu	nhỏ,	MSV_Scale_Out.jpg	nếu	phóng	to.	


**•** Trường	hợp	nhập	vào	tỷ	lệ	<100	ví	dụ	là	25:	

```python
# Nhập tỷ lệ thay đổi kích thước
scale = float(input("Nhập tỷ lệ thay đổi kích thước (%): "))

# Lấy kích thước ảnh gốc
h, w = img.shape[:2]

# Tính kích thước mới theo tỷ lệ
new_width = int(w * scale / 100)
new_height = int(h * scale / 100)

# Chọn thuật toán nội suy
if scale < 100:
        interpolation = cv2.INTER_NEAREST  # Thu nhỏ
        text = 'thu nhỏ' 
else:
        interpolation = cv2.INTER_LINEAR  # Phóng to
        text = 'phóng to'
# Thay đổi kích thước ảnh
resized_image = cv2.resize(img, (new_width, new_height), interpolation=interpolation)
print("--> bạn sẽ " ,text, "ảnh"  ) 
print("kích thước của ảnh gốc : " , h ,"x" , w) 
print("kích thước của ảnh sau khi thay đổi : " , new_height , "x" , new_width )
# Hiển thị ảnh gốc và ảnh sau khi thay đổi kích thước
plt.figure(figsize=(20,8))

plt.subplot(1,2,1)
plt.title('ảnh gốc ')
plt.imshow(img)


plt.subplot(1,2,2)
plt.title('ảnh đã scale')
plt.imshow(resized_image)
plt.show()

# Lưu ảnh đã thay đổi kích thước
save_path = 'Pic/Saves/2221050292_img_scale.jpg'
cv2.imwrite(save_path, resized_image)
```

**Yêu	cầu	7:**

**Thực	hiện	trên	nhiều	ảnh**


**1.** Sinh	viên	thực	hiện	thay	đổi	toàn	bộ	ảnh	trong	thư	mục	Pic_original	về	
ảnh	có	kích	thước	(width	=	128	pixel,	height	=	96	pixel).	Nếu	ngày	sinh	
<=15	lật	ảnh	theo	trục	X,	ngược	lại	lật	ảnh	theo	trục	Y.	

**2.** Lưu	ảnh	vào	thưc	mục	Pic_New	đặt	tên	ảnh	theo	quy	tắc	MSV_i.png	với	i	
là	số	thứ	tự	tăng	dần.	

**3.** Đọc	ảnh	màu	với	hệ	màu	RGB	và	Hiển	thị	toàn	bộ	ảnh	đã	xử	lý	trong	thư	
mục	Pic_New	ra	màn	hình.

```python
import cv2
import os
import matplotlib.pyplot as plt

# Nhập mã sinh viên
ma_sv = input("Nhập mã sinh viên: ")

# Nhập ngày sinh
ngay_sinh = int(input("Nhập ngày sinh: "))

# Thư mục chứa ảnh gốc và thư mục lưu ảnh mới
input_folder = "Pic/Pic_original"
output_folder = "Pic/Pic_New"

# Lấy danh sách ảnh trong thư mục Pic_original
image_files = [f for f in os.listdir(input_folder) if f.endswith((".jpg", ".png"))]

# Kiểm tra nếu thư mục không có ảnh
if not image_files:
    print("Không tìm thấy ảnh trong thư mục Pic_original!")
else:
    flipped_images = []

    for i, image_name in enumerate(image_files, start=1):
        image_path = os.path.join(input_folder, image_name)

        # Đọc ảnh
        image = cv2.imread(image_path)

        if image is None:
            print(f"Lỗi khi đọc ảnh: {image_name}")
            continue

        # Resize ảnh về 128x96
        image = cv2.resize(image, (128, 96))

        # Lật ảnh theo điều kiện ngày sinh
        if ngay_sinh <= 15:
            flipped_image = cv2.flip(image, 0)  # trục X
        else:
            flipped_image = cv2.flip(image, 1)  # trục Y

        # Lưu ảnh vào thư mục Pic_New
        output_image_name = f"{ma_sv}_{i}.png"
        output_path = os.path.join(output_folder, output_image_name)
        cv2.imwrite(output_path, flipped_image)

        flipped_images.append((flipped_image, output_image_name))

    # Hiển thị toàn bộ ảnh đã xử lý
    num_images = len(flipped_images)
    cols = 4
    rows = (num_images + cols - 1) // cols

    plt.figure(figsize=(15, 5))
    for idx, (img, name) in enumerate(flipped_images, start=1):
        plt.subplot(rows, cols, idx)
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        plt.title(name, fontsize=10)
        plt.axis("off")

    plt.tight_layout()
    plt.show()

    print(f"Đã xử lý và lưu {num_images} ảnh vào thư mục {output_folder}.")
```
