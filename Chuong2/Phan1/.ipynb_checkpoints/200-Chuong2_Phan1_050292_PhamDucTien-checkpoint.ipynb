### 1.1 Thay đổi kích thước ảnh (Resizing images)
---
* Thay đổi kích thước (Zoom in, Zoom out) là một trong những kỹ thuật xử lý ảnh thông dụng.
* Ảnh có kích thước lớn sẽ chiếm nhiều bộ nhớ, số lượng tập dữ liệu ảnh thường rất lớn. Do đó cần giảm kích thước ảnh. Resize là một kỹ thuật phổ biến. => Trong OpenCV, Sử dụng hàm cv2.resize()

```python
#Đọc và hiển thị ảnh gốc ban đầu: Apples.jpg
img_orignal = cv2.imread('images/Pic_1.jpg', cv2.IMREAD_COLOR_RGB)
if img_orignal is None:
    print('Lỗi đọc ảnh!')
else:
    plt.imshow(img_orignal)
    plt.title('Ảnh gốc')
    plt.show()


```

```python
#Lấy kích thước ảnh ban đầu:
h = img_orignal.shape[0]
w = img_orignal.shape[1]
print('Kích thước ảnh gốc:', h , 'x', w)
print('Số pixel của ảnh:', h*w,'pixel')
```

## Các tham số của phương thức resize:
---

**cv2.resize(src, dsize, interpolation)**

Ý nghĩa các tham số:

1. src: ảnh gốc cần thay đổi kích thước.
2. dsize (width, height): Kích thước ảnh muốn thay đổi
3. interpolation: chỉ định thuật toán thực hiện resize:
* INTER_NEAREST – nội suy láng giềng gần nhất
* INTER_LINEAR – nội suy song tuyến tính (mặc định)
* INTER_AREA – nội suy theo vùng.
* INTER_CUBIC –  nội suy xoắn bậc 3, trên  4×4 pixel láng giềng
* INTER_LANCZOS4 –  nội suy Lanczos trên 8×8 pixel láng giềng!


**Thông thường:**
* cv.INTER_AREA để thu nhỏ
* cv.INTER_CUBIC & cv.INTER_LINEAR để zoom.

```python
#Resize ảnh với các thuật toán khác nhau:
#1.Tăng kích thước của ảnh(Tăng độ phân giải)
h_new = 2000
w_new = 1200

dsize = (w_new, h_new)
img_tang = cv2.resize(img_orignal, dsize, cv2. INTER_CUBIC)

print('Số pixel ảnh mới', img_tang.shape[0]*img_tang.shape[1])
#Hiển thị kết quả:
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
plt.imshow(img_orignal)
plt.title('Ảnh gốc')

plt.subplot(1,2,2)
plt.imshow(img_tang)
plt.title('Ảnh tăng')
#lưu ảnh
cv2.imwrite('images/Pic_1_tang.jpg', img_tang)
```

```python
#Đổi hệ màu trước khi lưu

cv2.imwrite('images/Pic_1_tang.jpg', cv2.cvtColor(img_tang, cv2.COLOR_RGB2BGR))
```

```python
#2.Giảm độ phân giải (Thu nhỏ kích thước ảnh)
img_giam = cv2.resize(img_orignal,(64,96),cv2.INTER_AREA)
print('Số pixel ảnh mới', img_giam.shape[0]*img_giam.shape[1])
plt.figure(figsize = (12,5))
plt.subplot(1,2,1)
plt.imshow(img_orignal)
plt.title('Ảnh gốc')

plt.subplot(1,2,2)
plt.imshow(img_giam)
plt.title('Ảnh giảm')
```

## THỰC HÀNH 2.1 (Tiếp): 
---
Thay đổi kích thước nhiều ảnh:

**Yêu cầu 4:** Sinh viên sử dụng thư mục chứa các ảnh đã đổi tên thực hiện trong bài thực hành 1.4 chương 1. Thực hiện thay đổi kích thước toàn bộ ảnh về độ phân giải 128x128 pixel. Lưu lại ảnh đã thay đổi vào thư mục mới có tên Pic_128 và đặt lại tên ảnh như sau: Pic_128_i.jpg với i là các số thứ tự của ảnh.


```python
# {SINH VIÊN VIẾT CODE VÀO ĐÂY}
import cv2
import os

# Thư mục nguồn chứa ảnh gốc
input_folder = "images/Renamed/"

# Thư mục đích để lưu ảnh đã resize
output_folder = "Pic_128"

# Tạo thư mục nếu chưa tồn tại
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Lấy danh sách file ảnh trong thư mục nguồn
image_files = sorted(os.listdir(input_folder))  # Sắp xếp để đánh số thứ tự đúng

# Duyệt qua từng ảnh và resize
for i, filename in enumerate(image_files, start=1):
    input_path = os.path.join(input_folder, filename)  # Đường dẫn file gốc
    output_path = os.path.join(output_folder, f"Pic_128_{i}.jpg")  # Đường dẫn file mới

    # Đọc ảnh
    img = cv2.imread(input_path)

    if img is not None:  # Kiểm tra nếu ảnh hợp lệ
        # Resize về 256x256
        resized_img = cv2.resize(img, (256, 256))

        # Lưu ảnh đã resize vào thư mục Pic_256
        cv2.imwrite(output_path, resized_img)

print(" Resize hoàn tất! Ảnh đã lưu trong thư mục 'Pic_128'.")


```

## 2. Cắt ảnh (Cropping Images)
---
Thực hiện cắt một phần của ảnh, sử dụng phương pháp tách các phần tử trong ma trận:
* Ảnh nhị phân, ảnh xám --> Ma trận [hàng (chiều cao ảnh), côt (chiều rộng ảnh)]
* Ảnh màu ---> Mảng 3 chiều [hàng (chiều cao ảnh), côt (chiều rộng ảnh),màu (BGR)]

**img_cropped = img[start_row:end_row, start_col:end_col]**

```python
#Đọc và hiển thị ảnh màu: MissVN.jpg
img_crop = cv2.imread('images/MissVN.jpg',cv2.IMREAD_COLOR_RGB)
plt.imshow(img_crop)
plt.title('Ảnh Hoa Hậu Việt Nam')
plt.xticks(range(0, img_crop.shape[1], 20), rotation=90)
plt.yticks(range(0, img_crop.shape[0], 20))
plt.grid(ls=':')
plt.show()
```

```python
#Cắt lấy từng người trong ảnh gốc:
img_ahau2 = img_crop[50:,25:225]

img_hoahau = img_crop[10:, 225:425]

img_ahau1 = img_crop[25:,425:625]

plt.figure(figsize=(15,6))
plt.subplot(1,3,1)
plt.imshow(img_hoahau)
plt.title('Hoa hậu')

plt.subplot(1,3,2)
plt.imshow(img_ahau1)
plt.title('Á hậu 1')

plt.subplot(1,3,3)
plt.imshow(img_ahau2)
plt.title('Á hậu 2')

plt.show()


```

```python
#Sinh viên thực hiện Chỉ Cắt lấy khuôn mặt của từng người trong ảnh gốc và hiển thị kết quả
img_ahau2 = img_crop[40:200,60:200]

img_hoahau = img_crop[60:200, 260:360]

img_ahau1 = img_crop[40:180,460:560]

plt.figure(figsize=(15,6))
plt.subplot(1,3,1)
plt.imshow(img_hoahau)
plt.title('Hoa hậu')

plt.subplot(1,3,2)
plt.imshow(img_ahau1)
plt.title('Á hậu 1')

plt.subplot(1,3,3)
plt.imshow(img_ahau2)
plt.title('Á hậu 2')

plt.show()



```

```python
#CẮT ẢNH THEO MỘT VÙNG LỰA CHỌN BẰNG CHUỘT VỚI PHƯƠNG THỨC cv2.selectROI():
import cv2
import numpy as np
import matplotlib.pyplot as plt  
  
#Đọc ảnh:
img_miss = cv2.imread('images/MissVN.jpg')
  
# Select ROI: Lấy một vùng trong ảnh --> Toạ độ của vùng ảnh 
r = cv2.selectROI("select the area", img_miss)

print(r)

#Cắt ảnh theo vùng được chọn
cropped_image = img_miss[int(r[1]):int(r[1]+r[3]), 
                      int(r[0]):int(r[0]+r[2])]
  
# Hiển thị ảnh đã cắt:
plt.imshow(cv2.cvtColor(cropped_image,cv2.COLOR_BGR2RGB))
plt.grid(ls='--')
plt.show()
```

Hàm cv2.selectROI() trả về một tuple gồm 4 giá trị: (x,y,w,h)

* x: Tọa độ x của góc trên bên trái của hình chữ nhật (tính từ góc trái trên cùng của ảnh).
* y: Tọa độ y của góc trên bên trái của hình chữ nhật (tính từ góc trái trên cùng của ảnh).
* w: Chiều rộng (width) của hình chữ nhật.
* h: Chiều cao (height) của hình chữ nhật.

```python
#làm mờ một phần của ảnh được chọn:
#Đọc ảnh:
img_miss = cv2.imread('images/MissVN.jpg')
  
# Select ROI: Lấy một vùng trong ảnh --> Toạ độ của vùng ảnh 
r = cv2.selectROI("select the area", img_miss)

print(r)

#Đổi màu theo vùng được chọn
img_miss[int(r[1]):int(r[1]+r[3]), 
                      int(r[0]):int(r[0]+r[2])] = 100
  
# Hiển thị ảnh đã cắt:
plt.imshow(cv2.cvtColor(cropped_image,cv2.COLOR_BGR2RGB))
plt.grid(ls='--')
plt.show()
```

* Thực hiện xoay ảnh 90 độ thuận và ngược chiều kim đồng hồ.
* Thực hiện lật ảnh theo chiều ngang và dọc
* Hiển thị và lưu các ảnh đã xoay, lật vào thư mục images/Saves/
<img src = 'images/Thuchanh_2_3.jpg' width='300px'>

```python
#{SINH VIÊN VIẾT CODE VÀO ĐÂY}
#Đọc ảnh
import cv2
import matplotlib.pyplot as plt
img_th = cv2.imread('images/Thuchanh_2_3.jpg')
img_th23 = cv2.cvtColor(img_th, cv2.COLOR_BGR2RGB)
plt.imshow(img_th23)
plt.show()
```

```python
img_rotate90_th23 = cv2.rotate(img_th23, cv2.ROTATE_90_CLOCKWISE)
plt.imshow(img_rotate90_th23)
plt.show()

cv2.imwrite('images/Saves/img_rotate90_th23.jpg', img_rotate90_th23)
```

```python
img_rotate_90_th23_2 = cv2.rotate(img_th23, cv2.ROTATE_90_COUNTERCLOCKWISE)

#Hiển thị ảnh:

plt.imshow(img_rotate_90_th23_2 )
plt.show()

cv2.imwrite('images/Saves/img_rotate_90_th23_2 .jpg', img_rotate_90_th23_2)
```

```python
#Lật ảnh theo chiều ngang (trục X): flipCode = 0
img_flip_x = cv2.flip(img_th23,0)
#Hiển thị
plt.imshow(img_flip_x)
plt.show()

cv2.imwrite('images/Saves/img_flip_x.jpg', img_flip_x)
```
